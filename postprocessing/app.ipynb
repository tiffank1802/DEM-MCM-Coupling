```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
spectral_l_positions=pd.read_csv("https://huggingface.co/buckets/ktongue/DEM_MCM/resolve/_Good/Experiment/postraitement/spectral_simulations/spectral_5cells_vw0.5_k15_NLT30_step157_dt31_tau157_start250/fichiers/transitions/P_large_spectral_5cells_vw0.5_k15_NLT30_step157_dt31_tau157_start250.txt?download=true",sep='\s+',header=None)
spectral_s_positions=pd.read_csv("https://huggingface.co/buckets/ktongue/DEM_MCM/resolve/_Good/Experiment/postraitement/spectral_simulations/spectral_5cells_vw0.5_k15_NLT30_step157_dt31_tau157_start250/fichiers/transitions/P_small_spectral_5cells_vw0.5_k15_NLT30_step157_dt31_tau157_start250.txt?download=true",sep='\s+',header=None)
```

```python
spectral_l_positions=spectral_l_positions.to_numpy()
spectral_s_positions=spectral_s_positions.to_numpy()
```

```python
sum_l=spectral_l_positions.sum(axis=1)
sum_s=spectral_s_positions.sum(axis=1)
```

```python

# RSD
def rsd_stats(vec):
    std = vec.std()
    mean = vec.mean()
    rsd = std / mean
    return std, mean, rsd

std_l, mean_l, rsd_l = rsd_stats(sum_l)
std_s, mean_s, rsd_s = rsd_stats(sum_s)

# Figure (2,2)
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# (0,0) — Heatmap Large
sns.heatmap(spectral_l_positions, annot=True, fmt=".2f", cmap="Oranges",
            ax=axes[0, 0], linewidths=0.5)
axes[0, 0].set_title("P_large")

# (1,0) — Heatmap Small
sns.heatmap(spectral_s_positions, annot=True, fmt=".2f", cmap="Blues",
            ax=axes[1, 0], linewidths=0.5)
axes[1, 0].set_title("P_small")
# (0,1) — Somme par ligne + RSD pour Large
n = len(sum_l)
axes[0, 1].barh(range(n), sum_l, color="darkorange", alpha=0.7)
axes[0, 1].set_yticks(range(n))
axes[0, 1].set_yticklabels([f"État {i}" for i in range(n)])
axes[0, 1].axvline(x=1.0, color='red', linestyle='--', linewidth=1, label='Ref = 1')
axes[0, 1].set_title("Somme lignes — Large")
axes[0, 1].set_xlabel("Somme")
axes[0, 1].invert_yaxis()


# ✅ Valeurs sur les barres
for idx, val in enumerate(sum_l):
    axes[0, 1].text(val + 0.001, idx, f"{val:.4f}",
                    va='center', ha='left', fontsize=9, color='black')

textstr_l = f"mean = {mean_l:.4f}\nstd  = {std_l:.4f}\nRSD  = {rsd_l:.4f}"
axes[0, 1].text(0.05, 0.05, textstr_l, transform=axes[0, 1].transAxes,
                fontsize=10, verticalalignment='bottom',
                bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))
axes[0, 1].legend()

# (1,1) — Somme par ligne + RSD pour Small
axes[1, 1].barh(range(n), sum_s, color="steelblue", alpha=0.7)
axes[1, 1].set_yticks(range(n))
axes[1, 1].set_yticklabels([f"État {i}" for i in range(n)])
axes[1, 1].axvline(x=1.0, color='red', linestyle='--', linewidth=1, label='Ref = 1')
axes[1, 1].set_title("Somme lignes — Small")
axes[1, 1].set_xlabel("Somme")
axes[1, 1].invert_yaxis()


# ✅ Valeurs sur les barres
for idx, val in enumerate(sum_s):
    axes[1, 1].text(val + 0.001, idx, f"{val:.4f}",
                    va='center', ha='left', fontsize=9, color='black')

textstr_s = f"mean = {mean_s:.4f}\nstd  = {std_s:.4f}\nRSD  = {rsd_s:.4f}"
axes[1, 1].text(0.05, 0.05, textstr_s, transform=axes[1, 1].transAxes,
                fontsize=10, verticalalignment='bottom',
                bbox=dict(boxstyle='round', facecolor='moccasin', alpha=0.5))
axes[1, 1].legend()
```

## Influence du pas de temps tau sur la stabilité de la matrice de transition



**Utilisation de la matrice idéale**

L'opération vise à transformer la matrice empirique $M$ (`mat_52`) en une matrice idéale $M_{id}$ possédant deux propriétés fondamentales :

1.  **Diagonale uniforme** : Remplacer chaque auto-transition $M_{ii}$ par la moyenne globale $\bar{d}$ de la diagonale.
2.  **Stochasticité colonnaire** : Normaliser chaque colonne pour que $\sum_i (M_{id})_{ij} = 1$.

$$
\begin{cases}
(M')_{ij} = \begin{cases} 
\bar{d} = \frac{1}{n}\sum_{k=0}^{n-1} M_{kk} & \text{si } i=j \\
M_{ij} & \text{si } i \neq j 
\end{cases} \\[10pt]
(M_{id})_{ij} = \dfrac{(M')_{ij}}{\sum_{k=0}^{n-1} (M')_{kj}}
\end{cases}
$$
