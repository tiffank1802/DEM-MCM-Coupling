```python

# Passage en format long pour Seaborn
df_long = spectral.melt(id_vars='temps', var_name='Partition', value_name='Population')

# Tracé
fig, ax = plt.subplots(figsize=(12, 5))

sns.lineplot(
    data=df_long,
    x='temps',
    y='Population',
    hue='Partition',
    ax=ax
)

ax.set_title("Spectral 5 cells 'large' + 'small'")
ax.set_xlabel("Temps")
ax.set_ylabel("Population")
ax.legend(title="Partition", bbox_to_anchor=(1.01, 1), loc='upper left')

plt.tight_layout()
plt.show()
```
