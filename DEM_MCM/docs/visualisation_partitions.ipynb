```python
A=part.AdaptiveZPartitioner(bottom_method="cylindrical",bottom_kwargs={"nr":1,"ntheta":4,"nz":1})
```

```python
particles_nb=5000
theta = np.random.uniform(0, 2*np.pi, particles_nb)
r = np.random.uniform(0, 1, particles_nb)
z = np.random.uniform(0, 10, particles_nb)
x = r * np.cos(theta)
y = r * np.sin(theta)
coordinates = np.vstack([x, y, z]).T

# Créer le partitionneur

A.fit(coordinates)
    
# Visualiser automatiquement
A.visualize(x, y, z, 
                        plot_types=["3d", "2d_xy", "volumes"],
                        save_path="demo_partition")
```
